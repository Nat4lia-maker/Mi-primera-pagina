```python
import os

base = "mini_turtle_task"
os.makedirs(f"{base}/mini_turtle", exist_ok=True)

open(f"{base}/mini_turtle/__init__.py", "w").close()
open(f"{base}/mini_turtle/drawer_logic.py", "w").close()
open(f"{base}/main.py", "w").close()

print("Estructura creada ✔️")
```

```python
drawer_logic_code = """
posicion_x = 0

def _dibujar(marca="*"):
    print(" " * posicion_x + marca)

def adelante(pasos=1):
    global posicion_x
    posicion_x += pasos
    _dibujar("→")
    return posicion_x

def abajo():
    print()
    _dibujar("|")
    return posicion_x

def reiniciar():
    global posicion_x
    posicion_x = 0
    _dibujar("R")
    return posicion_x
"""

with open("mini_turtle_task/mini_turtle/drawer_logic.py", "w") as f:
    f.write(drawer_logic_code)

print("drawer_logic.py listo ✔️")
```

```python
init_code = """
from .drawer_logic import adelante, abajo, reiniciar

__all__ = ["adelante", "abajo", "reiniciar"]
"""

with open("mini_turtle_task/mini_turtle/__init__.py", "w") as f:
    f.write(init_code)

print("__init__.py listo ✔️")
```

```python
main_code = """
from mini_turtle import adelante, abajo, reiniciar

def dibujar_escalera(peldanos=4, paso=4):
    for _ in range(peldanos):
        adelante(paso)
        abajo()

def dibujar_rampa(pasos=5):
    for _ in range(pasos):
        adelante(2)

if __name__ == "__main__":
    print("=== ESCALERA ===")
    dibujar_escalera()

    print("\\n=== REINICIO ===")
    reiniciar()

    print("\\n=== RAMPA ===")
    dibujar_rampa()
"""

with open("mini_turtle_task/main.py", "w") as f:
    f.write(main_code)

print("main.py listo ✔️")
```
